```python
items_aggregated = items.groupby('order_id').agg(total_items=('order_item_id', 'max'),
                               no_unique_sellers = ('seller_id', 'nunique'),
                              latest_ship_date = ('shipping_limit_date', 'max'),
                              total_amount = ('price', 'sum'),
                               total_freight = ('freight_value', 'sum')).reset_index()
```
